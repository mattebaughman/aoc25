Part 1

```python
import numpy as np
import operator
from functools import reduce
```

```python
with open('input.txt','r') as f:
    data = f.read()
data=data.split('\n')[:-1]
```

```python
data = [k.split(' ') for k in data]
data = [[d for d in n if d] for n in data]
```

```python
def compute(d):
    ops={'+':operator.add,'*':operator.mul}
    return(reduce(ops[d[-1]],[int(f) for f in d[:-1]]))
```

```python
data = np.array(data).T
```

```python
total = 0
for d in data:
    total+=compute(d)
print(total)
```

Part 2

```python
with open('input.txt','r') as f:
    data = f.read()
data=data.split('\n')[:-1]
```

```python
splits = []
for d in range(len(data[0])):
    column_index = True
    for k in data:
        if k[d]!=' ':
            column_index = False
    if column_index:
        splits.append(d)
```

```python
parsed_data = []
for d in data:
    parsed_data.append([])
    ind=-1
    for s in splits:
        parsed_data[-1].append(d[ind+1:s])
        ind=s
    parsed_data[-1].append(d[ind+1:])
parsed_data = np.array(parsed_data)
```

```python
def new_number_generator(nums):
    a = np.array([np.array(list(k)) for k in nums])
    return(np.array([int(''.join(k)) for k in a.T]))
```

```python
ops={'+':operator.add,'*':operator.mul}
total = 0
for k in range(len(parsed_data[0])):
    problem = parsed_data[:,k]
    numbers = problem[:-1]
    prob_op = problem[-1][0]
    new_numbers = new_number_generator(numbers)
    total+=reduce(ops[prob_op],new_numbers)
print(total)
```
